### Working with dates - a worked example

This is an example of how you can work with dates in a different way from the one presented in Lecture 6.1. You will need it in Part 2. (You do not need to change this cell - just run it and study what each line does.)

```python
import pandas as pd

# Example DataFrame with a date column
example = pd.DataFrame({
    "Date": ["20200101", "20200201", "20200301", "20200401"],
    "Value": [10, 20, 30, 40],
})

# Convert 'Date' from text to dates (datetime format)
example["Date"] = pd.to_datetime(example["Date"], format="%Y%m%d")

# Extract the year, month and day from the 'Date' column
example["Year"] = example["Date"].dt.year
example["Month"] = example["Date"].dt.month
example["Day"] = example["Date"].dt.day

# Display the DataFrame with the new columns
print(example)
```
